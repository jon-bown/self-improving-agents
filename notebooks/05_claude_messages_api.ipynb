# Claude Messages API smoke test

Load `ANTHROPIC_API_KEY` from the project `.env` file and make one small request with the Anthropic Python SDK. Run the code cell to verify that the Messages API is working. The API key is never displayed.

```python
from pathlib import Path

import anthropic
from dotenv import dotenv_values

env_path = next(
    (directory / ".env" for directory in (Path.cwd(), *Path.cwd().parents)
     if (directory / ".env").is_file()),
    None,
)
if env_path is None:
    raise FileNotFoundError("Could not find a project .env file from the notebook's working directory.")

api_key = dotenv_values(env_path).get("ANTHROPIC_API_KEY")
if not api_key:
    raise RuntimeError(f"ANTHROPIC_API_KEY is missing or empty in {env_path}.")

client = anthropic.Anthropic(api_key=api_key)
response = client.messages.create(
    model="claude-haiku-5-5",
    max_tokens=64,
    messages=[{"role": "user", "content": "Reply with: Messages API is working."}],
)

text = "\n".join(block.text for block in response.content if block.type == "text")
print(text)
print(f"\nSuccess: {response.model} | input tokens: {response.usage.input_tokens} | output tokens: {response.usage.output_tokens}")
```
